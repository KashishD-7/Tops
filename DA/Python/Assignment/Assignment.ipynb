```python
#41. Write a Python program to check whether a list contains a sub list.

def sublist_container(main_list, sub_list):
  sub_len = len(sub_list)
  for i in range(len(main_list) - sub_len + 1):
    if main_list[i:i+sub_len] == sub_list:
      return True
  return False

#Take input from user
main = input("Enter Your main elements :").split()
sub = input("Enter Your sublist elements :").split()

# Validation
if sublist_container(main, sub):
  print("Main list Contains Sublist.")
else:
  print("Main list Does not Contains Sublist.")
```

```python
#61. Write a Python function to calculate the factorial of a number (a nonnegative integer)

def factorial(n):
    if n < 0:
        return "Factorial is not defined for negative numbers."
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

num = int(input("Enter a nonnegative integer: "))
print("Factorial:", factorial(num))
```

```python
#63. Write a Python function to check whether a number is perfect or not.
def is_perfect_number(n):
    # Negative numbers and 0 can't be perfect numbers
    if n <= 0:
        return False

    sum_of_divisors = 0  # Initialize sum of divisors

    # Loop from 1 to n-1 to find proper divisors
    for i in range(1, n):
        if n % i == 0:      # If i divides n completely
            sum_of_divisors += i  # Add it to the sum
    return sum_of_divisors == n

num = int(input("Enter a number: "))

if is_perfect_number(num):
    print(f"{num} is a perfect number.")
else:
    print(f"{num} is not a perfect number.")
```

```python
#64. Write a Python function that checks whether a passed string is palindrome or not

def is_palindrome(s):
    # Remove spaces and convert to lowercase for uniform comparison
    s = s.replace(" ", "").lower()

    # Compare original with reversed string
    return s == s[::-1]

user_input = input("Enter a string: ")

if is_palindrome(user_input):
    print("Yes It is a palindrome.")
else:
    print("No It is not a palindrome.")
```

```python
#80. Write a Python program to count the frequency of words in a file.

word_freq = {}

# Open the file in read mode
with open("example.txt", "r") as file:
    for line in file:
        words = line.strip().split()  # Split line into words
        for word in words:
            word = word.lower().strip('.,!?()"\'')  # Normalize the word
            if word:
                word_freq[word] = word_freq.get(word, 0) + 1

print("Word Frequencies:")
for word, freq in word_freq.items():
    print(f"{word}: {freq}")
```
